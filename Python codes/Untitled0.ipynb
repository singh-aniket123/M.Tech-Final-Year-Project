```python
import pandas as pd

# Sample DataFrame creation (assuming df already exists and has 'timestamp' column)
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Failure Report
failures = [
    {'start': '2020-04-18 00:00:00', 'end': '2020-04-18 23:59:00'},
    {'start': '2020-05-29 23:30:00', 'end': '2020-05-30 06:00:00'},
    {'start': '2020-06-05 10:00:00', 'end': '2020-06-07 14:30:00'},
    {'start': '2020-07-15 14:30:00', 'end': '2020-07-15 19:00:00'}
]

# Convert failure report times to datetime
failures = [{'start': pd.to_datetime(f['start']), 'end': pd.to_datetime(f['end'])} for f in failures]

def calculate_hours_till_failure(timestamp):
    # Initialize a large number for minimum hours
    min_hours = float('inf')

    # Convert timestamp to datetime
    timestamp = pd.to_datetime(timestamp)

    for failure in failures:
        if failure['start'] <= timestamp <= failure['end']:
            return 0  # Inside a failure period, so hours till failure is 0

        # Calculate time difference from the start of the failure period
        if timestamp < failure['start']:
            # Calculate the number of hours till the failure period starts
            delta = failure['start'] - timestamp
            hours_till_failure = delta.total_seconds() / 3600 # Convert seconds to hours
            min_hours = min(min_hours, hours_till_failure)

    return min_hours if min_hours != float('inf') else 0 # Return 0 if outside or after failure period beacuse failure period is already recorded.

# Apply the function to each row in the dataframe
df['Hours_till_Failure'] = df['timestamp'].apply(calculate_hours_till_failure)
df.head(10)
```

```python
import pandas as pd

# Assuming your dataset is in a DataFrame called df

# Convert 'timestamp' column to datetime type if it's not already
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Set 'timestamp' as the DataFrame index
df.set_index('timestamp', inplace=True)

# Define the aggregation methods for each column
aggregation_methods = {
    'TP2': 'mean',
    'TP3': 'mean',
    'H1': 'mean',
    'DV_pressure': 'mean',
    'Reservoirs': 'mean',
    'Oil_temperature': 'mean',
    'Motor_current': 'mean',
    'COMP': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'DV_eletric': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'Towers': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'MPG': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'LPS': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'Pressure_switch': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'Oil_level': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'Caudal_impulses': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'Severity': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty
    'Failure': lambda x: x.mode()[0] if not x.mode().empty else None, # Return None if mode is empty  # Most frequent value
    'Hours_till_Failure': 'mean'  # Assuming numeric
}

# Resample to 1-minute frequency and apply aggregation function
df_resampled = df.resample('2T').agg(aggregation_methods)

# If needed, reset the index to turn the timestamp back into a column
df_resampled.reset_index(inplace=True)

# Display resampled data
df_resampled.head(10)
```
